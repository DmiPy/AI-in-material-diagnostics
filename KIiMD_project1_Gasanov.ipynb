```python
import numpy as np, matplotlib.pyplot as plt, librosa as lib, os
```

```python
folder = 'KIiMD_project1'
file_names = sorted([f for f in os.listdir(folder) if f.endswith('.wav')])
paths = []
sounds = []

for f in file_names:
    new_elem = os.path.join(folder, f)
    paths.append(new_elem)
    y, sr = lib.load(new_elem)
    sounds.append(y)
```

```python
parts = [f.split('_') for f in file_names]
print(parts)
```

```python
md = []
for j in range(len(file_names)):
     md.append({'fn':file_names[j], 'sig': sounds[j], 'spec': parts[j][0], 'mID': parts[j][4], 
            'time': parts[j][5], 'rid': parts[j][6], 'sID': (parts[j][-1].split('.'))[0]})
print(md[0])
```

```python
def plot_spectrogramm(signal, title):
    plt.specgram(signal, cmap='magma'
                        #, NFFT=None, Fs=None, detrend=None, sides=None,
                        #xextent=None, mode=None, scale=None
                       )
    #plt.xlabel()
    #plt.ylabel()
    plt.title(title)
    #plt.show()

```

```python
for i in range(len(md)):
    signal1 = np.fft.fft(md[i]['sig'])
    md[i].update({'stft': signal1})
```
